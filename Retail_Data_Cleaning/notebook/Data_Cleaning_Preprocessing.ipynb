```python
# Step 12 — Standardize text columns

text_columns = [
    "customer_segment",
    "city",
    "category",
    "payment_status"
]

for col in text_columns:
    orders[col] = orders[col].astype("string").str.strip()
```

```python
# Step 14 — Handle missing values

orders.isna().sum()
```

```python
#Quantity

orders["quantity"] = orders["quantity"].fillna(
    orders["quantity"].median()
)
```

```python
# Unit Price

orders["unit_price"] = orders["unit_price"].fillna(
    orders["unit_price"].median()
)
```

```python
# Discount

orders["discount_pct"] = orders["discount_pct"].fillna(0)
```

```python
# categorical values

orders["customer_segment"] = orders["customer_segment"].fillna(
    "Unknown"
)

orders["city"] = orders["city"].fillna(
    "Unknown"
)

orders["category"] = orders["category"].fillna(
    "Unknown"
)

orders["payment_status"] = orders["payment_status"].fillna(
    "Unknown"
)
```

```python
# For Dates

orders["order_date"] = orders["order_date"].fillna(
    orders["order_date"].median()
)
```

```python
# Step 15 — Check missing values after cleaning

orders.isna().sum()
```

```python
# Step 16 — Remove duplicate records

before_duplicates = len(orders)

orders = orders.drop_duplicates()

after_duplicates = len(orders)

print("Rows before:", before_duplicates)
print("Rows after:", after_duplicates)
print("Duplicates removed:", before_duplicates - after_duplicates)
```

```python
# Step 17 — Check outliers

orders[["quantity", "unit_price", "discount_pct"]].describe()
```

```python
#Quantity outliers

#Calculate IQR:

Q1 = orders["quantity"].quantile(0.25)
Q3 = orders["quantity"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

quantity_outliers = orders[
    (orders["quantity"] < lower_bound) |
    (orders["quantity"] > upper_bound)
]

print("Quantity outliers:", len(quantity_outliers))
```

```python
# Step 18 — Handle unrealistic values

#Quantity
#Quantity should not be negative

(orders["quantity"] < 0).sum()
```

```python
orders = orders[orders["quantity"] >= 0]
```

```python
# Unit Price
(orders["unit_price"] < 0).sum()
```

```python
# Remove invalid negative prices:
orders = orders[orders["unit_price"] >= 0]
```

```python
# Discount

invalid_discount = (
    (orders["discount_pct"] < 0) |
    (orders["discount_pct"] > 100)
)

invalid_discount.sum()
```

```python
# Handled them

orders.loc[
    orders["discount_pct"] < 0,
    "discount_pct"
] = 0

orders.loc[
    orders["discount_pct"] > 100,
    "discount_pct"
] = 100
```

# Step 19 — Feature Engineering

This is specifically required for our  assignment.

We will create:

Year
Month
Month Name
Gross Sales
Discount Amount
Net Sales


```python
# Extract year

orders["year"] = orders["order_date"].dt.year
```

```python
# Extract month

orders["month"] = orders["order_date"].dt.month
```

```python
# Extract month name

orders["month_name"] = (
    orders["order_date"].dt.month_name()
)
```

```python
# Step 20 — Calculate Gross Sales

#Formula -->(Gross Sales = Quantity × Unit Price)

orders["gross_sales"] = (
    orders["quantity"] *
    orders["unit_price"]
)
```

```python
# Step 21 — Calculate Discount Amount

# Formula -->(Discount Amount = Gross Sales × Discount % / 100)

orders["discount_amount"] = (
    orders["gross_sales"] *
    orders["discount_pct"] / 100
)
```

```python
# Step 22 — Calculate Net Sales

#Formula --> (Net Sales = Gross Sales - Discount Amount)

orders["net_sales"] = (
    orders["gross_sales"] -
    orders["discount_amount"]
)
```
